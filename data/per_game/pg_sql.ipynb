## Cleaning Data


```python
# Get number of null values in each column
null_counts = df_drop_cols.null_count()
null_counts
```

```python
df_fill_null = df_drop_cols.fill_null(0)
```

```python
# Get number of null values in each column
df_fill_null = df_fill_null.with_columns([pl.col('home').cast(pl.Int64)])
df_fill_null.head()
```

```python
master = df_fill_null.select([
    "game_id",
    "season",
    "playerteamcity",
    "opponentteamcity"])
# Drop duplicate rows from master_with_features
master = master.unique()
master.shape
```

```python
stat_cols = [
    "numminutes", "points", "assists", "blocks", "steals", "fieldgoalsattempted", "fieldgoalsmade", "fieldgoalspercentage", "threepointersattempted", "threepointersmade", "threepointerspercentage", "freethrowsattempted", "freethrowsmade", "freethrowspercentage", "reboundsdefensive", "reboundsoffensive", "reboundstotal", "foulspersonal", "turnovers", "per", "ts_percent", "x3p_ar", "f_tr", "trb_percent", "ast_percent", "stl_percent", "blk_percent", "tov_percent", "pg_x3p_percent", "x2p_per_game", "x2pa_per_game", "pg_x2p_percent", "pg_e_fg_percent", "ft_per_game", "fta_per_game", "pg_ft_percent", "trb_per_game", "ast_per_game", "stl_per_game", "blk_per_game", "tov_per_game", "pf_per_game", "pts_per_game", "shoot_fg_percent", "avg_dist_fga", "percent_fga_from_x2p_range", "percent_fga_from_x0_3_range", "percent_fga_from_x3_10_range", "percent_fga_from_x10_16_range", "percent_fga_from_x16_3p_range", "percent_fga_from_x3p_range", "fg_percent_from_x2p_range", "fg_percent_from_x0_3_range", "fg_percent_from_x3_10_range", "fg_percent_from_x10_16_range", "fg_percent_from_x16_3p_range", "fg_percent_from_x3p_range", "percent_assisted_x2p_fg", "percent_assisted_x3p_fg", "percent_dunks_of_fga", "num_of_dunks", "percent_corner_3s_of_3pa", "corner_3_point_percent"
]
for stat in stat_cols:
    for t in ["t0", "t1"]:
        for pNum in range(15):
                col_name = f"{t}_p{pNum}_{stat}"
                master = master.with_columns([
                    pl.lit(0).alias(col_name)
                ])

master.head()
```

```python

# Step 3: Rank by numMinutes within each (gameId, teamIndex)
features_ranked = df_fill_null.with_columns([
    pl.col("numminutes").rank("dense", descending=True)
    .over(["game_id", "home"])
    .alias("playerrank")
])

# Step 4: Keep top 15 players per team
top_players = features_ranked.filter(pl.col("playerrank") <= 15)

# Step 5: Reshape (melt → pivot → wide format)
stat_cols = [
    "numminutes", "points", "assists", "blocks", "steals", "fieldgoalsattempted", "fieldgoalsmade", "fieldgoalspercentage", "threepointersattempted", "threepointersmade", "threepointerspercentage", "freethrowsattempted", "freethrowsmade", "freethrowspercentage", "reboundsdefensive", "reboundsoffensive", "reboundstotal", "foulspersonal", "turnovers", "per", "ts_percent", "x3p_ar", "f_tr", "trb_percent", "ast_percent", "stl_percent", "blk_percent", "tov_percent", "pg_x3p_percent", "x2p_per_game", "x2pa_per_game", "pg_x2p_percent", "pg_e_fg_percent", "ft_per_game", "fta_per_game", "pg_ft_percent", "trb_per_game", "ast_per_game", "stl_per_game", "blk_per_game", "tov_per_game", "pf_per_game", "pts_per_game", "shoot_fg_percent", "avg_dist_fga", "percent_fga_from_x2p_range", "percent_fga_from_x0_3_range", "percent_fga_from_x3_10_range", "percent_fga_from_x10_16_range", "percent_fga_from_x16_3p_range", "percent_fga_from_x3p_range", "fg_percent_from_x2p_range", "fg_percent_from_x0_3_range", "fg_percent_from_x3_10_range", "fg_percent_from_x10_16_range", "fg_percent_from_x16_3p_range", "fg_percent_from_x3p_range", "percent_assisted_x2p_fg", "percent_assisted_x3p_fg", "percent_dunks_of_fga", "num_of_dunks", "percent_corner_3s_of_3pa", "corner_3_point_percent"
]

melted = top_players.melt(
    id_vars=["game_id", "home", "playerrank"],
    value_vars=stat_cols
).with_columns([
    (pl.lit("t") + pl.col("home").cast(pl.Utf8) +
     pl.lit("_p") + (pl.col("playerrank") - 1).cast(pl.Utf8) +
     pl.lit("_") + pl.col("variable")).alias("column_name")
])

wide = melted.pivot(
    values="value",
    index="game_id",
    columns="column_name",
    aggregate_function="first"
)
# Step 6: Join back to master
master_with_features = master.update(wide, on="game_id")
master_with_features.head()
```

```python
# Get number of rows where t0_p0_numminutes == 0
num_zero_minutes = master_with_features.filter(pl.col('t0_p0_numminutes') == 0).height
num_zero_minutes
```

```python
import itertools
from tqdm import tqdm

# Step 1: For each game and team, get top 15 players ordered by minutes
grouped = top_players.groupby(['gameId', 'teamIndex'])
permuted_rows = []
for (gameId, teamIndex), group in tqdm(grouped):
    players = group.sort('numminutes', reverse=True).to_dicts()
    # Generate all permutations of player orderings
    for perm in itertools.permutations(players, len(players)):
        row = {'gameId': gameId, 'teamIndex': teamIndex}
        for i, player in enumerate(perm):
            for stat in stat_cols:
                col_name = f't{teamIndex}_p{i}_{stat}'
                row[col_name] = player.get(stat, None)
        permuted_rows.append(row)

# Step 2: Create DataFrame from permuted rows
permuted_df = pl.DataFrame(permuted_rows)
permuted_df.head()
```
